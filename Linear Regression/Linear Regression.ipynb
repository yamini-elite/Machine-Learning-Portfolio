```python
import pandas as pd

from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

data = {
    "Age":[1,2,3,4,5,6,7],
    "Price":[950000,880000,810000,720000,620000,510000,430000]
}

df = pd.DataFrame(data)

X = df[["Age"]]
y = df["Price"]

poly = PolynomialFeatures(degree=2)

X_poly = poly.fit_transform(X)

model = LinearRegression()

model.fit(X_poly, y)

new_car = pd.DataFrame({
    "Age":[8]
})

new_car_poly = poly.transform(new_car)

prediction = model.predict(new_car_poly)

print(prediction)
```

```python
import pandas as pd
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

data={
    "Age":[1,2,3,4,5,6,7],
    "Price":[950000,880000,810000,720000,620000,510000,430000]
}
df = pd.DataFrame(data)
X = df[["Age"]]
Y = df["Price"]
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)
model = LinearRegression()
model.fit(X_poly,Y)
new_car = pd.DataFrame({
    "Age":[10]
})

new_car_poly = poly.transform(new_car)
prediction = model.predict(new_car_poly)
print(prediction)
```

exp vs salary

```python
import pandas as pd
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
data = {
    "exp":[1,2,3,4,5,6,7],
    "salary":[25000,32000,42000,55000,70000,85000,98000]
}
df = pd.DataFrame(data)
X = df[["exp"]]
Y = df["salary"]
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)
model = LinearRegression()
model.fit(X_poly,Y)
new_salary = pd.DataFrame({
    "exp":[8]
})
new_poly=poly.transform(new_salary)
prediction = model.predict(new_poly)
print(prediction)
```

Ridge Regression

```python
import pandas as pd
from sklearn.linear_model import Ridge

data = {
    "Experience":[1,2,3,4,5,6,7,8],
    "Salary":[25000,32000,42000,55000,70000,85000,98000,110000]
}

df = pd.DataFrame(data)
X = df[["Experience"]]
Y =df["Salary"]
model = Ridge(alpha=100)
model.fit(X,Y)
new_data = pd.DataFrame({
    "Experience":[9]
})
prediction = model.predict(new_data)
print(prediction)
```

LASSO


```python
import pandas as pd
from sklearn.linear_model import Lasso
from sklearn.model_selection import train_test_split
from sklearn.metrics import (mean_absolute_error,
    mean_squared_error,
    r2_score)
import numpy as np
```

```python
df =pd.read_csv("/content/house_price.csv")
df.isnull().sum()
```

```python
X = df[["Area",
        "Bedrooms",
        "House_Age",
        "Distance_City"]]

Y = df["Price"]
```

```python
X_train,X_test,Y_train,Y_test =train_test_split(X,Y,test_size = 0.2,random_state=42)
```

```python
model = Lasso(alpha =100)
model.fit(X_train,Y_train)
y_pred = model.predict(X_test)
```

```python
mae = mean_absolute_error(Y_test, y_pred)

mse = mean_squared_error(Y_test, y_pred)

rmse = np.sqrt(mse)

r2 = r2_score(Y_test, y_pred)

print("MAE :", mae)

print("MSE :", mse)

print("RMSE :", rmse)

print("R2 :", r2)
```

```python
print(model.coef_)
```

```python
new_house = pd.DataFrame({
    "Area":[1800],
    "Bedrooms":[4],
    "House_Age":[5],
    "Distance_City":[3]
})

price = model.predict(new_house)

print(price)
```
